```python
import astropy
from astropy.table import Table
import astropy.units as u
import numpy as np
import os, sys
```

```python
inroot = 'WHL0137_weight5_MSA17'
inroot = 'WHL0137_MSA'

if 0:
    infile = inroot + '.cat'
    obsfile = 'MSAobs1.txt'
    outfile = inroot + '_exposure2.cat'
else:
    infile = inroot + '_exposure2.cat'
    obsfile = 'MSAobs2.txt'
    outfile = inroot + '_exposure3.cat'

print(infile)
print(obsfile)
print(outfile)
```

```python
MSA_catalog = Table.read(infile, format='ascii.commented_header')
MSA_catalog[:2]
```

```python
observed_targets = Table.read(obsfile, format='ascii.tab')
observed_targets[:3]
```

```python
n = len(observed_targets)
n
```

```python
for iobs in range(n):
    id = observed_targets['id'][iobs]
    if id == 10000: # Earendel
        pass  
    elif id < 20000: # not background
        icat = list(MSA_catalog['id']).index(id)
        MSA_catalog['weight'][icat] /= 100.
```

```python
MSA_catalog.write(outfile, format='ascii.commented_header', delimiter=' ', overwrite=True)
MSA_catalog
```
